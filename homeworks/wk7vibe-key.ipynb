## Who wants to share their answers?

```python
# 1. Which borough had the most summons? 

boro_counts = df["BORO"].value_counts()
boro_counts
```

```python
## write as an f-string
# Count summons by borough (sorted highest to lowest automatically)

# Get the top two boroughs
top_two = boro_counts.head(2)

# Extract the name of the borough with the most summons
first_boro = top_two.index[0]

# Extract the number of summons for that borough
first_count = top_two.iloc[0]

# Extract the name of the borough with the second-most summons
second_boro = top_two.index[1]

# Extract the number of summons for that borough
second_count = top_two.iloc[1]

# Create a formatted sentence using f-strings
# :, adds comma formatting to large numbers
# .title() makes borough names title case instead of ALL CAPS
summary = (
    f"{first_boro.title()} issued the most summons "
    f"({first_count:,}), followed by {second_boro.title()} "
    f"with {second_count:,}."
)

# Print the final sentence
print(summary)
```

```python
# 2. What were the top 10 most common offenses?
# Get top 10 offenses
top10 = (
    df["OFFENSE_DESCRIPTION"]
    .value_counts()
    .head(10)
)

print(top10)
```

```python
# 3. What are the total number of summons broken down by year?
# Convert to datetime
df["year"] = pd.to_datetime(df["SUMMONS_DATE"], errors="coerce")
summons_by_year = (
    df["year"]
    .dt.year
    .value_counts()
    .sort_index()
)

print(summons_by_year)
```

```python
# 4. What were some of the most unusual summons?
pd.set_option("display.max_colwidth", None)
# Add a length column
df["desc_length"] = df["OFFENSE_DESCRIPTION"].str.len()

# Sort by longest descriptions
df.sort_values("desc_length", ascending=False)[
    "OFFENSE_DESCRIPTION"
].head(10)
```

```python
# 6. How many summons were issued by race?
df["RACE"].value_counts()
```

```python
#7. Which borough had the fewest offenses per 100K people?


# summons by borough
borough_counts = df["BORO"].value_counts()

# 2020 Census population
populations = {
    "BROOKLYN": 2736074,
    "QUEENS": 2405464,
    "MANHATTAN": 1694251,
    "BRONX": 1472654,
    "STATEN ISLAND": 495747
}

pop = pd.Series(populations)

# combine + compute rate
per_capita = pd.concat([borough_counts, pop], axis=1)
per_capita.columns = ["summons_count", "population"]

per_capita["summons_per_100k"] = (
    per_capita["summons_count"] / per_capita["population"] * 100000
)

per_capita.sort_values("summons_per_100k")
```
